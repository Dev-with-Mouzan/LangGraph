# Types of Memory in LangGraph

LangGraph primarily supports **two types of memory**:

## 1. Short-Term Memory

Short-term memory stores information **within a single conversation or workflow thread**. It remembers the current context while the graph is running or across multiple turns of the same thread.

It is typically implemented using:
- **State**
- **Checkpointer**
- **Thread ID**

### Example

```text
User: My name is Alice.

User: What's my name?

AI: Your name is Alice.
```

Since both messages belong to the same thread, the graph remembers the user's name.

**Use Cases**
- Chatbots
- Multi-turn conversations
- Human-in-the-loop workflows
- Long-running agents

---

## 2. Long-Term Memory

Long-term memory stores information **across different conversations or threads**. It allows the graph to remember facts even after a conversation has ended.

It is typically implemented using:
- Databases
- Vector stores
- Knowledge bases
- External storage systems

### Example

```text
Conversation 1

User: I like Python.
```

Days later:

```text
Conversation 2

AI: I remember that your favorite programming language is Python.
```

The information is retrieved from long-term memory.

**Use Cases**
- Personalized AI assistants
- Customer support systems
- User preference storage
- Recommendation systems

---

# Comparison

| Feature | Short-Term Memory | Long-Term Memory |
|---------|-------------------|------------------|
| Scope | Current thread | Across multiple threads |
| Lifetime | Temporary | Persistent |
| Storage | Graph state + Checkpointer | External database or vector store |
| Uses Thread ID | ✅ Yes | ❌ Not necessarily |
| Best For | Conversations and workflows | User profiles and knowledge |
