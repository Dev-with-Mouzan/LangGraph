# What is CRAG (Corrective Retrieval-Augmented Generation)?

**CRAG (Corrective Retrieval-Augmented Generation)** is an advanced RAG technique that **evaluates the quality of retrieved documents before generating a response**. If the retrieved information is insufficient or irrelevant, CRAG retrieves additional information (e.g., from the web or another data source) before generating the final answer.

In simple terms:

> **CRAG verifies retrieved documents and corrects poor retrieval before the LLM generates an answer.**

---

## Why Do We Need CRAG?

In a traditional RAG system:

```text
User Query
     │
     ▼
Retrieve Documents
     │
     ▼
LLM Generates Answer
```

If the retrieved documents are incorrect or incomplete, the LLM may produce an inaccurate answer.

CRAG solves this by adding a validation step.

---

## CRAG Workflow

```text
User Query
     │
     ▼
Retrieve Documents
     │
     ▼
Evaluate Retrieved Documents
     │
 ┌───┴────────────┐
 ▼                ▼
Good Docs     Poor Docs
 │                │
 ▼                ▼
Generate      Retrieve More
Answer        Information
 │                │
 └───────┬────────┘
         ▼
   Generate Answer
```

---

## Example

### User Query

```text
What are the latest features of LangGraph?
```

### Retrieved Documents

- Old documentation (2023)

CRAG evaluates the documents and determines they are outdated.

It then:

- Retrieves updated documentation or web results.
- Combines the new information.
- Generates a more accurate answer.

---

## Benefits

- Improves answer accuracy
- Detects poor retrieval results
- Reduces hallucinations
- Retrieves additional information when needed
- Produces more reliable responses

---

## Use Cases

- AI search assistants
- Enterprise knowledge bases
- Customer support bots
- Research assistants
- Question-answering systems

---

## Traditional RAG vs CRAG

| Traditional RAG | CRAG |
|-----------------|------|
| Uses retrieved documents directly | Evaluates retrieved documents first |
| May answer with poor context | Corrects poor retrieval before answering |
| Higher chance of hallucinations | More accurate and reliable responses |
| Simpler pipeline | Adds a retrieval quality evaluation step |
