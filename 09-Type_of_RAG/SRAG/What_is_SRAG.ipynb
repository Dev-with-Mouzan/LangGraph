# What is SRAG (Self-RAG)?

**Self-RAG (Self-Reflective Retrieval-Augmented Generation)** is an advanced RAG technique where the **LLM evaluates its own retrieval process and generated responses**, deciding whether additional retrieval is needed to improve the answer.

In simple terms:

> **Self-RAG allows the LLM to "think before answering" by deciding when to retrieve more information and by verifying its own response.**

---

## Why Do We Need Self-RAG?

In traditional RAG:

```text
User Query
     │
     ▼
Retrieve Documents
     │
     ▼
Generate Answer
```

The model always uses the retrieved documents, even if they are insufficient.

Self-RAG adds self-evaluation:

- Are the retrieved documents relevant?
- Do I have enough information?
- Should I retrieve more documents?
- Is my generated answer reliable?

---

## Self-RAG Workflow

```text
User Query
     │
     ▼
Retrieve Documents
     │
     ▼
LLM Self-Evaluation
     │
 ┌───┴────────────┐
 ▼                ▼
Enough Info   Need More Info
 │                │
 ▼                ▼
Generate      Retrieve Again
Answer            │
                  ▼
          Generate Answer
```

---

## Example

### User Query

```text
Explain the latest LangGraph features.
```

The LLM retrieves documentation and evaluates it.

- If the information is sufficient → Generate the answer.
- If the information is incomplete or outdated → Retrieve more documents before answering.

---

## Benefits

- Improves answer quality
- Reduces hallucinations
- Retrieves additional information only when necessary
- Performs self-verification before responding
- Produces more reliable answers

---

## Use Cases

- AI research assistants
- Enterprise search
- Customer support systems
- Knowledge assistants
- Complex question-answering applications

---

## RAG vs CRAG vs Self-RAG

| Feature | RAG | CRAG | Self-RAG |
|---------|-----|------|----------|
| Retrieves documents | ✅ | ✅ | ✅ |
| Evaluates retrieved documents | ❌ | ✅ | ✅ |
| Retrieves additional information if needed | ❌ | ✅ | ✅ |
| LLM evaluates its own answer | ❌ | ❌ | ✅ |
| Reduces hallucinations | Moderate | High | Very High |
