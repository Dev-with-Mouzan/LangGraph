# Memory in LangGraph

**Memory** in LangGraph is the ability of a graph to **store and recall information across multiple executions or conversations**. It allows the graph to remember previous interactions instead of starting with an empty state every time.

Memory is typically implemented using:
- **Checkpointer** → Saves and loads the graph's state.
- **Thread ID** → Identifies which conversation or workflow state to retrieve.

## Workflow

```text
User Request
     │
     ▼
Load Memory
     │
     ▼
Execute Graph
     │
     ▼
Update State
     │
     ▼
Save Memory
```

## Common Use Cases

- Chatbots with conversation history
- Personalized AI assistants
- Long-running workflows
- Multi-turn question answering

## Benefits

- Remembers previous interactions
- Enables personalized responses
- Supports long-term conversations
- Allows workflows to resume from the last state
